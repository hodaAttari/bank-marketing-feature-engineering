```python
%pip install flaml
%pip install "flaml[automl]"

import pandas as pd
from flaml import AutoML
from sklearn.metrics import f1_score
```

```python
df = pd.read_csv('bank-additional-full.csv', na_values='unknown')
df.head()
```

    <br>
    If executing this cell fails, it could be due to one of the following reasons:
</font>
</p>

<div dir=ltr style="direction: ltr;text-align: left;line-height:200%;font-family:vazir;font-size:medium">

```python
target_variable = 'y'
df_train = df.sample(frac=.7, random_state=313)

model = AutoML(task='classification', time_budget=120, verbose=0, estimator_list=['lgbm', 'rf'])
model.fit(df_train.drop(target_variable, axis=1), df_train[target_variable])

df_test = df.drop(df_train.index)
y_pred = model.predict(df_test.drop(target_variable, axis=1))

f1score = f1_score(df_test[target_variable], y_pred, pos_label='yes')*100
print(f'performance of model is {f1score} %')
```

<h2 align=left style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
Part 1
</font>
</h2>

<p dir=ltr style="direction: ltr;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
As mentioned earlier, there are missing values in the categorical columns of this dataframe. As the first part of this question, we.</li>
    <li>In the <code>binary_cols</code> list, write the names of the columns that have binary discrete values.</li>
    <li>In the <code>nominal_cols</code> list, write the names of the columns that have nominal values.</li>
    <li>In the <code>ordinal_cols</code> list, write the names of the columns that have ordinal values.</li>
</ul>
</font>
</p>

```python
# TO-DO: fill `categorical_cols` and fill na values in categorical columns of `df` dataframe.

categorical_cols = [col for col in df.columns if df[col].dtype == 'object' and col != target_variable]

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

df.head()
```

<h2 align=left style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
Part 2
</font>
</h2>

<p dir=ltr style="direction: ltr;text-align: left;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
In this part, we expect you to prepare 4 lists. In each list, the columns of the <code>df</code> dataframe are grouped according to their type.

<ul dir=ltr>
    <li>In the <code>continuous_cols</code> list, write the names of the columns that have continuous values.</li>
    <li>In the <code>binary_cols</code> list, write the names of the columns that have binary discrete values.</li>
    <li>In the <code>nominal_cols</code> list, write the names of the columns that have nominal values.</li>
    <li>In the <code>ordinal_cols</code> list, write the names of the columns that have ordinal values.</li>
</ul>
</font>
</p>

<p dir=ltr style="direction: ltr;text-align: left;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
Note that the type (<code>dtype</code>) of all variables introduced above must be a list.
</font>
</p>


```python
continuous_cols = ['age', 'duration', 'campaign', 'pdays', 'previous']
binary_cols = ['default', 'housing', 'loan', 'is_telephone_contact']
nominal_cols = ['job', 'marital', 'month', 'day_of_week', 'poutcome']
ordinal_cols = ['education']
```


<h2 align=left style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
Part 3
</font>
</h2>

<p dir=ltr style="direction: ltr;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
In the third part, we want you to have a dataframe that only contains the binary columns of the <code>df</code> dataframe. Then, convert all values of the new dataframe into numbers.
<br>
You can build the <code>binary_df</code> dataframe using <code>binary_cols</code>.
Actually, the <code>binary_df</code> dataframe is a part of the <code>df</code> dataframe that only includes its binary columns. That is, no other column with continuous, nominal, or ordinal values should exist in the <code>binary_df</code> dataframe. Note that all values in this dataframe must be numbers (replace <code>yes</code> with <code>1</code> and <code>no</code> with <code>0</code>).
</font>
</p>

```python
# TO-DO: create `binary_df` and replace all `object` values with numeric ones
binary_df = df[binary_cols].copy()

# Convert yes/no to 1/0
for col in ['default', 'housing', 'loan','is_telephone_contact']:
    binary_df[col] = binary_df[col].map({'yes': 1, 'no': 0})

binary_df.head()
```


<h2 align=left style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
Part 4
</font>
</h2>

<p dir=ltr style="direction: ltr;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
In the fourth part, we want you to have a dataframe that only contains the nominal columns of the <code>df</code> dataframe.
<br>
Build the <code>nominal_df</code> dataframe using <code>nominal_cols</code>.
Since there is no order in nominal features, they must be One-Hot encoded.
The One-Hot encoding method for nominal discrete columns should be done such that the names of the created columns have two parts: the first part is the <b>name of the column</b> being encoded, and the second part is the name of the <b>nominal value</b>, separated by an underscore (<code>_</code>).
For a better understanding, consider the following example.
Assume there was a hypothetical column named <code>imaginary_col</code> in the dataset that had nominal values as follows:
</font>
</p>

<center>
<div style="direction: ltr;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
        
|<code>imaginary_col</code> |
| :---: |
|<code>val1</code>|
|<code>val1</code>|
|<code>val2</code>|
|<code>val3</code>|
|<code>val2</code>|
|<code>val1</code>|
|<code>val3</code>|

</font>
</div>
</center>

<p dir=ltr style="direction: ltr;text-align: left;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
    The result you should prepare for this column will look like this:
</font>
</p>

<center>
<div style="line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
        
|<code>imaginary_col_val1</code>|<code>imaginary_col_val2</code>|<code>imaginary_col_val3</code>|
| :---: | :---: | :---: |
|<code>1</code>|<code>0</code>|<code>0</code>|
|<code>1</code>|<code>0</code>|<code>0</code>|
|<code>0</code>|<code>1</code>|<code>0</code>|
|<code>0</code>|<code>0</code>|<code>1</code>|
|<code>0</code>|<code>1</code>|<code>0</code>|
|<code>1</code>|<code>0</code>|<code>0</code>|
|<code>0</code>|<code>0</code>|<code>1</code>|

</font>
</div>

<p dir=ltr style="direction: ltr;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
Repeat this for every column in the <code>df</code> dataframe that is of nominal type.
Finally, concatenate the resulting table from One-Hot encoding each column and store the final result in the <code>nominal_df</code> variable.
Note that the <code>nominal_df</code> dataframe should not include the original columns that were One-Hot encoded. Only include the encoded columns in this dataframe.
</font>
</p>


```python
nominal_df = pd.get_dummies(
    df[nominal_cols],
    prefix=nominal_cols,
    prefix_sep='_',
    drop_first=False,
    dtype=int
)

nominal_df.head()
```


<h2 align=left style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
Part 5
</font>
</h2>

<p dir=ltr style="direction: ltr;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazirmatn" size=3>
In the fifth part, we want you to have a dataframe that only contains the ordinal columns of the <code>df</code> dataframe. Then, convert all values of the new dataframe to numbers.
<br>
Build the <code>ordinal_df</code> dataframe using <code>ordinal_cols</code>.
As you did in the third part for binary columns, do the same for the ordinal columns. That is, convert the values in the <code>ordinal_df</code> dataframe to numbers using the mapping table provided at the beginning of the exercise.
</font>
</p>

```python
# PART 5 — create `ordinal_df` according to above explanations

# Define the correct order for the ordinal column 'education'
education_order = {
    'unknown': -1,
    'illiterate': 0,
    'basic.4y': 1,
    'basic.6y': 2,
    'basic.9y': 3,
    'high.school': 4,
    'professional.course': 5,
    'university.degree': 6
}

# Create ordinal_df containing only ordinal columns
ordinal_df = df[ordinal_cols].copy()

# Map textual values to numeric order
ordinal_df['education'] = ordinal_df['education'].map(education_order)

ordinal_df.head()
```

```python
target_variable = 'y'
new_df = df[continuous_cols].join([binary_df, nominal_df, ordinal_df])

model = AutoML(task='classification', time_budget=120, verbose=0, estimator_list=['lgbm', 'rf'])
model.fit(new_df, df[target_variable])

df_test = new_df.join(df.y).sample(frac=0.3, random_state=313)
y_pred = model.predict(df_test.drop(target_variable, axis=1))

f1score = f1_score(df_test[target_variable], y_pred, pos_label='yes')*100
print(f'performance of model is {f1score} %')
```
